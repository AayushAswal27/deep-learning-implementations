```python
# tokenize 
def tokenize(text):
    text=text.lower()
    text = text.replace("?", " ")
    text = text.replace("'", " ")
    return text.split()
    
```

```python
tokenize("What is the capital of France?")
```

```python
tokenize("Who wrote 'To Kill a Mockingbird'?")
```
